```python
# 5. Practical Exercise: Cleaning a Messy Customer Dataset

# Step 1: Import Pandas and load the dataset

import pandas as pd

df = pd.read_csv(r"C:\Users\HP\Downloads\Internship\week_2_day_3\details.csv")
print(df.head())

# Step 2: Check missing values

print(df.isnull().sum())

# Step 3: Check and remove duplicates
print("duplicates")
print(df.duplicated().sum())
df = df.drop_duplicates()

# Step 4: Correct data types

df["age"] = pd.to_numeric(df["age"], errors="coerce")
print(df.dtypes)
# Step 5: Handle missing values

df["age"] = df["age"].fillna(df["age"].median())
df["name"] = df["name"].fillna("Unknown")
df["city"] = df["city"].fillna("Unknown")

print(df.isnull().sum())
# Step 6: Identify outliers in age

Q1 = df["age"].quantile(0.25)
Q3 = df["age"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["age"] < lower) | (df["age"] > upper)]

print("Potential outliers:")
print(outliers)
# Step 7: Encode the city column

df = pd.get_dummies(df, columns=["city"], dtype=int)

print(df.head())
# Step 8: Save the cleaned dataset

df.to_csv(
    r"C:\Users\HP\Downloads\Internship\week_2_day_3\cleaned_details.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```

```python
import pandas as pd

data = {
    "customer_id": [101, 102, 103, 103, 104, 105, 106],
    "name": ["Ravi", "Priya", None, None, "Arjun", "Sneha", "Kiran"],
    "age": [25, 30, 28, 28, "unknown", 150, 35],
    "city": ["Nellore", "Chennai", "Hyderabad",
             "Hyderabad", "Nellore", None, "Chennai"],
    "purchase_amount": [2000, 3500, 1500, 1500, 2500, 999999, 4000]
}

df = pd.DataFrame(data)

df.to_csv("messy_customers.csv", index=False)

print(df)
print("printing the values before we the datset")
before_rows = len(df)
before_duplicates = df.duplicated().sum()
before_missing = df.isnull().sum().sum()

print("BEFORE CLEANING")
print("Total rows:", before_rows)
print("Duplicate rows:", before_duplicates)
print("Missing values:", before_missing)
print("\nMissing values per column:")
print(df.isnull().sum())
print("\nData types:")
print(df.dtypes)
print("removin the duplicates")
df = df.drop_duplicates()

print("Duplicates remaining:", df.duplicated().sum())
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["purchase_amount"] = pd.to_numeric(
    df["purchase_amount"], errors="coerce"
)
print(df.dtypes)
for column in ["age", "purchase_amount"]:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) | (df[column] > upper)
    ]

    print("\nOutliers in", column)
    print(outliers[["customer_id", column]])
df["age"] = df["age"].fillna(df["age"].median())
df["name"] = df["name"].fillna("Unknown")
df["city"] = df["city"].fillna("Unknown")

print(df.isnull().sum())
df = pd.get_dummies(df, columns=["city"], dtype=int)

print(df.head())
df.to_csv("cleaned_customers.csv", index=False)

print("Cleaned dataset saved successfully.")
after_rows = len(df)
after_duplicates = df.duplicated().sum()
after_missing = df.isnull().sum().sum()

report = pd.DataFrame({
    "Metric": [
        "Total rows",
        "Duplicate rows",
        "Missing values"
    ],
    "Before cleaning": [
        before_rows,
        before_duplicates,
        before_missing
    ],
    "After cleaning": [
        after_rows,
        after_duplicates,
        after_missing
    ]
})

print(report)
report.to_csv("before_after_report.csv", index=False)
```

1). What is data cleaning?

Data cleaning is the process of identifying and correcting missing, duplicate, invalid, or inconsistent data to improve data quality.

2). Why do we handle missing values?

Missing values can affect calculations and analysis. We may fill them, remove records when appropriate, or leave them missing depending on the situation.

3). What are duplicate records?

Duplicate records are repeated records in a dataset. Removing unintended duplicates helps prevent incorrect counts and analysis.

4). What is a data type?

A data type defines the kind of value stored in a column, such as an integer, decimal number, or text.

5). What is an outlier?

An outlier is a value that differs considerably from the other values. It should be investigated because it may be an error or a genuine unusual observation.

6). What is encoding?

Encoding converts categorical data into a numerical representation that is useful for many data-analysis and machine-learning tasks.

7). What is the purpose of a before/after report?

It shows how the dataset changed during cleaning, including changes in missing values, duplicate rows, and record counts.
