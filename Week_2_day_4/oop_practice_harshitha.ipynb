```python
# 4. Part B – Constructors and Attributes
# A constructor is commonly written using __init__(). It runs when an object is created and can assign initial values to that object.
# •	Create a class called Mobile with brand, model, and price attributes.
class Mobile:

# •	Use __init__() to initialize all three attributes.
    def __init__(self,brand,model,price):
        self.brand=brand
        self.model=model
        self.price=price
    
# •	Create at least three Mobile objects with different values.
m1=Mobile("samsung","A55",35000)
m2=Mobile("Apple","iphone15",65000)
m3=Mobile("Xiaomi","Redmi Note 13",24000)
# •	Print the details of every object.
print(m1.brand,m1.model,m1.price)
print(m2.brand,m2.model,m2.price)
print(m3.brand,m3.model,m3.price)
# Add a new attribute called storage and display it.
m1.storage="high storage"
m2.storage="low storage"
m3.storage="medium storage"
print(m1.brand,m1.model,m1.price,m1.storage)
print(m2.brand,m2.model,m2.price,m2.storage)
print(m3.brand,m3.model,m3.price,m3.storage)
# Try creating an object without passing a required value. Observe the error and explain why it happens.
mobile1=Mobile()
print(mobile1.brand,mobile1.model)
#TypeError: Mobile.__init__() missing 3 required positional arguments: 'brand', 'model', and 'price'
```

```python
# 5. Part C – Methods
# Methods are functions defined inside a class. They commonly use self to access the current object's attributes.
# •	Create a class called BankAccount with account_holder and balance.
class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance
# •	Add a display_balance() method.
    def display_balance(self):
        print(self.account_holder)
        print(self.balance)
# •	Add a deposit(amount) method that increases the balance.
    def deposit(self,amount):
        if amount<0:
            print("Negative depost is not allowed")
        else:
            self.balance=self.balance+amount
            print("balance after deposit",self.balance)
# •	Add a withdraw(amount) method that decreases the balance only when enough money is available.
    def withdraw(self,amount):
        if amount<=self.balance:
            self.balance=self.balance-amount 
            print("balnce after withdraw",self.balance)
        else:
            print("insufficient balance")
# •	Create an object and test deposit and withdrawal operations.
b1=BankAccount("Harshitha",20000)
b1.deposit(2000)
b1.withdraw(2000)
# Challenge: Add validation so a negative deposit is rejected. Also handle a withdrawal request that is greater than the current balance.
b1.deposit(-1000)
b1.withdraw(200000000)
```

```python
# 6. Part D – Methods That Return Values
# A method can print a result or return a value using return. Practice returning values from methods.
# •	Create a Rectangle class with length and width.
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width
# •	Add an area() method that returns length multiplied by width.
    def area(self):
        return self.length*self.width
        
# •	Add a perimeter() method that returns 2 multiplied by the sum of length and width.
    def perimeter(self):
        return 2*(self.length+self.width)
       
# •	Create two Rectangle objects and print their areas and perimeters.
r1=Rectangle(5,4)
r2=Rectangle(10,12)
area1=r1.area()
print(area1)
perimeter1=r1.perimeter()
print(perimeter1)
area2=r2.area()
print(area2)
perimeter2=r2.perimeter()
print(perimeter2)
```

```python
# 7. Part E – Inheritance Overview
# Inheritance allows a child class to reuse attributes and methods from a parent class. A child class can also define its own methods.
# •	Create a parent class called Person with name and age.
# •	Add a introduce() method that prints the person's details.
# •	Create a child class called Employee that inherits from Person.
# •	Add an employee_id attribute to Employee.
# •	Create an Employee object and call both the inherited method and a method of its own.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
 
    def introduce(self):
        print("Name:", self.name)
        print("Age:", self.age)
 
class Employee(Person):
    def __init__(self, name, age, employee_id):
        super().__init__(name, age)
        self.employee_id = employee_id
 
    def show_employee_id(self):
        print("Employee ID:", self.employee_id)
 
employee = Employee("Ananya", 23, "E101")
employee.introduce()
employee.show_employee_id()

# •	Create another child class called Trainer with a subject attribute.
class Trainer(Person):
    def __init__(self,name,age,subject):
        super().__init__(name,age)
        self.subject=subject
    def display(self):
        print("subject:",self.subject)
t1=Trainer("harshitha",21,"python")
t1.display()
t1.introduce()
# •	Try calling a method that exists only in the child class from a parent-class object. Record what happens.
p1=Person("harshitha",21)
p1.show_employee_id()
#In person parent class there is no show_employee_id() method
```

```python
# 10. Part H – Main Hands-on Activity: Student Record System
# Build a simple class-based Student Record System. The program should create student objects, store their details, and display useful information.
# Each student should have:
# •	name
# •	roll_number
# •	marks
# Your program should:
# •	Create a Student class with a constructor
# •	Add a display_details() method
# •	Add a result() method that returns Pass if marks are 40 or above and Fail otherwise
# •	Create at least four student objects with different marks
# •	Display each student's details and result
# •	Find and display the student with the highest marks
# •	Handle or prevent invalid marks outside the range 0–100
# •	Use a list to store the student objects
class Student:
    def __init__(self, name, roll_number, marks):
        # Marks validation
        if marks < 0 or marks > 100:
            raise ValueError("Marks must be between 0 and 100")
        self.name = name
        self.roll_number = roll_number
        self.marks = marks
 
    def display_details(self):
        print(self.name, self.roll_number, self.marks)
 
    def result(self):
        if self.marks >= 40:
            return "Pass"
        return "Fail"
 
students = [
    Student("Rahul", 1, 85),
    Student("Priya", 2, 92),
    Student("Arjun", 3, 38),
    Student("Ananya", 4, 76)
]
 
for student in students:
    student.display_details()
    print("Result:", student.result())
highest_student = students[0]

for student in students:
    if student.marks > highest_student.marks:
        highest_student = student

print("Student with Highest Marks")
print("Name:", highest_student.name)
print("Roll Number:", highest_student.roll_number)
print("Highest Marks:", highest_student.marks)
class GradeStudent(Student):
    def __init__(self,name,roll_number,marks,subject):
        super().__init__(name,roll_number,marks)
        self.subject=subject
g1=GradeStudent("Harshitha",4358,98,"java")
g1.display_details()
print(g1.marks)
```
