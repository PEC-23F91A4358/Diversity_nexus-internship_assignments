```python
# 1.	Load the data with pandas. Display the first five rows, shape, data types, missing-value counts, duplicate count, and target distribution. 
import pandas as pd

df = pd.DataFrame({
    "age": [24, 35, 29, 42, 31, None, 27, 39, 33, 46, 28, 36],
    "monthly_spend": [500, 1200, None, 2200, 900, None,
                      750, 1800, 1100, 2500, 650, 1500],
    "city": ["Hyderabad", "Chennai", "Hyderabad", "Bengaluru",
             "Chennai", "Hyderabad", "Visakhapatnam", "Bengaluru",
             "Chennai", "Hyderabad", "Bengaluru", "Chennai"],
    "plan_type": ["Basic", "Premium", "Standard", "Premium",
                  "Basic", "Basic", "Standard", "Premium",
                  "Standard", "Premium", "Basic", "Standard"],
    "support_tickets": [2, 0, 1, 3, None, 4, 1, 2, 0, 5, 1, 2],
    "churn": ["Yes", "No", "No", "Yes", "No", "Yes",
              "No", "No", "No", "Yes", "No", "Yes"]
})

display(df.head())
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate count:", df.duplicated().sum())
print("\nTarget distribution:")
print(df["churn"].value_counts())
# 2.Identify the target and input features. Explain why the target must not be included in X. 
X = df[[
    "age",
    "monthly_spend",
    "city",
    "plan_type",
    "support_tickets"
]]

y = df["churn"]
# 3.	Split into train and test sets before fitting any imputer, scaler, or encoder. Use a fixed random_state and explain the purpose. (5 marks)/
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)
# 4.	Build a numeric pipeline with median imputation and scaling. (7 marks)
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numeric_features = [
    "age",
    "monthly_spend",
    "support_tickets"
]

numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
# 5.	Build a categorical pipeline with missing-value imputation and OneHotEncoder(handle_unknown='ignore'). (7 marks)
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    "city",
    "plan_type"
]

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
# 6.	Combine the steps using ColumnTransformer. Fit on training data only and transform both train and test data. (6 marks)
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])
# 7.	Print transformed shapes and explain what changed. Identify one possible leakage risk. (5 marks)
# Fit on training data and transform it
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using the fitted preprocessor
X_test_processed = preprocessor.transform(X_test)

# Print shapes
print("Original training shape:", X_train.shape)
print("Original testing shape:", X_test.shape)

print("Transformed training shape:", X_train_processed.shape)
print("Transformed testing shape:", X_test_processed.shape)
```
