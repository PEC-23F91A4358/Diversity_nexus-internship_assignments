```python
import matplotlib.pyplot as plt

semesters = ["1st Sem", "2nd Sem", "3rd Sem", "4th Sem", "5th Sem"]
marks = [75, 82, 78, 88, 90]

plt.plot(semesters, marks, marker="o")

plt.xlabel("Semester")
plt.ylabel("Marks")
plt.title("Student Marks Across Semesters")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Rahul", "Priya", "Arjun", "Sneha"]
marks = [84, 90, 68, 95]

plt.bar(students, marks, label="Marks")

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Comparison of Student Marks")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45, 52, 56, 61, 65, 68, 72, 75, 81, 85, 89, 92]

plt.hist(marks, bins=5, edgecolor="black")

plt.xlabel("Marks Range")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [40, 50, 55, 65, 75, 85]

plt.scatter(study_hours, marks, label="Students")

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.legend()
plt.show()
```

. Explain the concepts in your own words

1. What is Matplotlib?
   Matplotlib is a Python library used to create graphs and charts. It helps us understand and visualize data.
2. What is a line chart?
   A line chart connects data points with lines. It helps us understand changes over time .
3. What is a bar chart?
   A bar chart uses rectangular bars to compare values between different categories.
4. What is a histogram?
   A histogram groups numerical data into intervals and shows the frequency of values in each interval.
5. What is a scatter plot?
   A scatter plot uses dots to show the relationship between two numerical variables.
6. What are labels?
   Labels explain what the x-axis and y-axis represent.
7. What is a legend?
   A legend identifies the data series shown in a chart.
8. Why do we use plt.show()?
   We use plt.show() to display the completed graph
