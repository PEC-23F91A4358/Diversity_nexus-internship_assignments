```python

# 4. Part B – Read a CSV File
# Create students.csv with:
# name,age,marks,city
# Rahul,21,85,Hyderabad
# Priya,22,92,Bangalore
# Arjun,20,78,Chennai
#Open the CSV file.
import csv
with open("students.csv","w",newline="") as file:
    writer=csv.writer(file)
    writer.writerow(["name","age","marks","city"])
    writer.writerow(["Rahul",21,85,"Hyderabad"])
    writer.writerow(["Priya",22,92,"Bangalore"])
    writer.writerow(["Arjun",20,78,"Chennai"])
#Read each record.
with open("students.csv","r") as file:
    data=csv.reader(file)
    for i in data:
       print(i)
    
# Print each student's name and marks.
with open("students.csv","r") as file:
    data=csv.reader(file)
    next(data)
    for i in data:
        print(i[0] ,i[2])
       
# Find the average marks.
with open("students.csv","r") as file:
    data=csv.reader(file)
    next(data)
    total=0
    count=0
    for i in data:
        total=total+int(i[2])
        count=count+1
    avg_marks=total/count
    print(avg_marks)
        
# Find the student with the highest marks.
with open("students.csv","r") as file:
    data=csv.reader(file)
    next(data)
    highest_marks=0
    highest_student=""
    for i in data:
        if highest_marks<int(i[2]):
            highest_marks=int(i[2])
            highest_student=i[0]
    print(highest_student,highest_marks)
    



```

```python
# 10. Part H – Main Hands-on Activity: Data Processing Script
# Build a simple data processing script that reads student records, cleans the data, handles bad records, and exports the cleaned result.
# Use a CSV file such as:
# name,age,marks,city
# Rahul,21,85,Hyderabad
# Priya,22,92,Bangalore
# ,20,75,Chennai
# Arjun,abc,80,Chennai
# Sneha,21,95,
# Kiran,20,88,Hyderabad
import csv
with open("details.csv","w",newline="") as file:
    writer=csv.writer(file)
    writer.writerow(["name","age","marks","city"])
    writer.writerow(["Rahul",21,85,"Hyderabad"])
    writer.writerow(["Priya",22,92,"Bangalore"])
    writer.writerow(["",20,75,"Chennai"])
    writer.writerow(["Arjun","abc",80,"chennai"])
    writer.writerow(["Sneha",21,95,""])
    writer.writerow(["Kiran",20,88,"Hyderabad"])
# Your program should:
# •	Read the CSV file.
with open("details.csv","r") as file:
    data=csv.reader(file)
    for i in data:       
        print(i)
# •	Check for missing names or cities.
with open("details.csv","r") as file:
    data=csv.reader(file)
    next(data)
    for i in data:
        if i[0]=="":
           
           print("invalid :missing name")
        if i[3]=="":
            print("invalid :missing city")
# •	Handle invalid age/marks values without crashing.
        try:
           age=int(i[1])
           marks=int(i[2])
        except ValueError:
            print("in valid age or marks")
# •	Clean valid records.
valid_records=[]
invalid_records=[]
with open("details.csv","r") as file:
    data=csv.reader(file)
    next(data)
    for i in data:
        if i[0]=="" or i[3]=="":
            invalid_records.append(i)
            continue
        try:
            age=int(i[1])
            marks=int(i[2])
        except ValueError:
            print("in valid age or marks")
            continue
        valid_records.append([i[0],age,marks,i[3]])
    print("valid records:")
    print(valid_records)
    print("in valid records:")
    print(invalid_records)
    for record in invalid_records:
       print(record)

# •	Ignore or separately report invalid records.
print("in valid records:")
for record in invalid_records:
       print(record)

# •	Calculate the average marks of valid students.
total = 0

for student in valid_records:
    total = total + student[2]

average = total / len(valid_records)

print("Average marks:", average)
# •	Find the highest scorer.
highest_marks = 0
highest_student = ""

for student in valid_records:
    if student[2] > highest_marks:
        highest_marks = student[2]
        highest_student = student[0]

print("Highest scorer:", highest_student)
print("Highest marks:", highest_marks)
# •	Export valid records into cleaned_students.csv.
with open("cleaned_students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "age", "marks", "city"])

    writer.writerows(valid_records)
# •	Save invalid records into errors.csv or display them clearly.
with open("errors.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "age", "marks", "city"])

    writer.writerows(invalid_records)
```
